```python
from nsetools import Nse
import random
import pandas as pd

nse = Nse()

# Get all NSE stock codes
all_stocks = list(nse.get_stock_codes().keys())[1:]  # skip header
print("Total Stocks:", len(all_stocks))

# Pick 20 random stocks for today
random_stocks = random.sample(all_stocks, 20)
print("Today's Random 20 Stocks:", random_stocks)

# Save to CSV (optional)
pd.DataFrame(random_stocks, columns=['Ticker']).to_csv("data/todays_random_stocks.csv", index=False)
```
