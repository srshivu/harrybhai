```python
from ultralytics import YOLO
import cv2

# Load YOLOv8 food model (trained or downloaded)
model = YOLO("yolov8n_food.pt")   # You can use yolov8n.pt for testing

# Load an image
image_path = "images/food.jpg"
image = cv2.imread(image_path)

# Run YOLOv8
results = model(image)

# Print results
for result in results:
    for box in result.boxes:
        cls = int(box.cls[0])
        conf = float(box.conf[0])
        x1, y1, x2, y2 = box.xyxy[0]

        print("Food:", model.names[cls])
        print("Confidence:", round(conf, 2))
        print("Bounding Box:", (x1, y1, x2, y2))
        print("--------")

# Show output image
annotated = results[0].plot()
cv2.imshow("Detected Food", annotated)
cv2.waitKey(0)
```
