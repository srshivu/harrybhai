```python
code = 200
match code:
    case 200:
        print("ok")
    case 404:
        print("not found")
    case 500:
        print("server error")
    case _:
        print("unknown")   # default case
```

```python
def httpstatus(status):
    match status:
        case 200:
            return "ok"
        case 400:
            return "server not found"
        case 500:
            return "not found"
        case _:
            print("default function")
httpstatus(300)            
```
