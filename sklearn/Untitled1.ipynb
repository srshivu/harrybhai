```python
bitcode = int(input("Enter a number: "))
s = []

# Divide by 2 until number becomes 0
while bitcode > 0:
    remainder = bitcode % 2
    s.append(remainder)
    bitcode//=2

# Reverse the list to get correct binary order
s.reverse()

# Print binary number
print("Binary:", ''.join(map(str, s)))
```
