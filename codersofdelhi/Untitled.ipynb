```python
def datainfo(fe):
    print("Users and their connections")

    # Create lookup dictionaries for fast ID → name mapping
    user_dict = {user["id"]: user["name"] for user in fe["users"]}
    page_dict = {page["id"]: page["name"] for page in fe["pages"]}

    for user in fe["users"]:
        # Convert friend IDs into friend names
        friends = ", ".join(user_dict[fid] for fid in user["friends"])#if shiva.strip=treu else if empty .strip false thats onluy you need to go

        # Convert liked page IDs into page names
        liked_pages = ", ".join(page_dict[pid] for pid in user["liked_pages"])

        print(f"The {user['name']} is friends with {friends} and likes pages {liked_pages}")

    print("\nPages information")
    for page in fe["pages"]:
        print(f"{page['id']} : {page['name']}")
```

```python
import json
def clean_data(data):
    data["users"]=[user for user in data["users"] if user["name"].strip()]
    for user in data["users"]:
        user["friends"]=list(set(user["friends"]))
    data["users"]=[user for user in data["users"] if user["friends"] or user["liked_pages"]]
    return data
data=json.load(open("data2.json"))
clean_data(data)
json.dump(data,open("cleaned_data.json","w"),indent=4)
print("the data it will cleaned sucessufully")
data["users"]
```
