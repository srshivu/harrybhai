```python
import matplotlib.pyplot as plt

players = ["Sachin", "Sehwag", "Kohli"]
runs = [1500, 1200, 1800]
explode=[0.1,0,0]


# Create a pie chart
plt.pie(runs, labels=players, autopct="%1.1f%%", startangle=90, colors=["skyblue", "lightgreen", "orange"],explode=explode)

plt.title("Runs Distribution by Players")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Data
num3 = np.arange(2001, 2011)  # numbers from 2001 to 2010

# Random colors
colors = np.random.rand(len(num3), 3)
print(colors)

# Explode effect (random values between 0 and 0.3)
explode = np.random.uniform(0, 0.3, size=len(num3))

# Plot pie chart
plt.pie(num3, labels=num3, autopct='%2.1f%%', colors=colors, explode=explode)

plt.show()
```
