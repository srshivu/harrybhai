```python
import matplotlib.pyplot as plt
years = [1990, 1992, 1994, 1996, 1998, 2000, 2003, 2005, 2007, 2010]
runs =  [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500]

plt.plot(years, runs)
plt.show()
```

```python

kohli = [0, 0, 500, 800, 1100, 1300, 1500, 1800, 1900, 2100]
sehwag = [0, 300, 800, 1200, 1500, 1700, 1600, 1400, 1000, 0]

plt.plot(years, kohli, label="Virat Kohli")
plt.plot(years, sehwag, label="Virender Sehwag")

plt.xlabel("Year")
plt.ylabel("Runs Scored")
plt.title("Performance Comparison")
plt.legend(["virat","sewhag"])#legend has more privority
plt.show()
```

```python
plt.plot(years, kohli, 'b*--', label="Kohli")  # red circles with dashed lines
plt.plot(years, sehwag, 'y^-', label="Sehwag")  # green triangles dotted
plt.legend()
```

```python
plt.plot(years,kohli,color="blue",linestyle='--',label="kohli era")
plt.plot(years,sehwag,color="red",linestyle="dotted",label="sewhag era")
plt.plot(years,runs,linestyle="-",label="sachine")
plt.grid()
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.style.use('Solarize_Light2')   # Apply style
plt.plot(x, y)
plt.title("Sine Wave with ggplot style")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y = np.sin(x)

# Enable XKCD style
plt.xkcd()

plt.plot(x, y, label="Sine Wave")
plt.title("This looks like XKCD style!")
plt.legend()
plt.show()
```

```python
with plt.xkcd():
    plt.bar([1, 2, 3], [3, 5, 7])
    plt.title("Temporary XKCD Bar Plot")
    plt.show()
```
