```python
import numpy as np

sachin = [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500]
sehwag = [0, 200, 900, 1400, 1600, 1800, 1500, 1100, 800, 0]
kohli  = [0, 0, 500, 800, 1100, 1300, 1500, 1800, 1900, 2100]

x = np.arange(len(years))  # index positions
width = 0.25
print(x)
```

```python
print(x-width)
```

```python
print(x+width)
```

```python
plt.bar(x - width, sachin, width=width, label="Sachin")
plt.bar(x, sehwag, width=width, label="Sehwag")
plt.bar(x + width, kohli, width=width, label="Kohli")

plt.xlabel("Year")
plt.ylabel("Runs")
plt.title("Run Comparison")
plt.xticks(x, years)  # Show actual year instead of 0,1,2,...
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Categories
categories = ["A", "B", "C", "D"]

# Two groups of values
group1 = [4, 7, 2, 9]
group2 = [5, 6, 3, 8]

# Create positions on X-axis
x = np.arange(len(categories))   # [0,1,2,3]
width = 0.25
# width of each bar

# Plot bars
plt.bar(x - width/2, group1, width, label="Group 1", color="skyblue")
plt.bar(x + width/2, group2, width, label="Group 2", color="orange")

# Labels  # set category names
plt.xlabel("Categories")
plt.ylabel("Values")
plt.title("Side by Side Bar Chart")
plt.xticks(x,categories)  # Show actual year instead of 0,1,2,...
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Dataset
categories = ["A", "B", "C", "D"]
group1 = [20, 34, 30, 35]
group2 = [25, 30, 28, 32]
group3 = [18, 28, 25, 30]

x = np.arange(len(categories))   # positions
width = 0.11                # width of each bar

# Plot 3 groups side by side
plt.bar(x-width,group1,width,color="red")
plt.bar(x,group2,width)
plt.bar(x+width,group3,width,color="green")
```

```python
import pandas as pd
data={"sachin" : [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500],
"sehwag" :[0, 200, 900, 1400, 1600, 1800, 1500, 1100, 800, 0],
"kohli"  : [0, 0, 500, 800, 1100, 1300, 1500, 1800, 1900, 2100]
     ,"years" : [1990, 1992, 1993, 1996, 1998, 2000, 2003, 2005, 2007, 2010]}

df=pd.DataFrame(data)
df["runs"]=df["sachin"]+df["sehwag"]+df["kohli"]
x=np.arange(len(df["years"]))
width=0.25
plt.bar(x-width,df["sachin"],width,color="red")
plt.bar(x,df["sehwag"],width,color="blue")
plt.bar(x+width,df["kohli"],width)
```
