```python
import seaborn as sns
flights=sns.load_dataset("flights")
```

```python
flights
```

```python
import pandas as pd
pv=flights.pivot(index="month",columns="year",values="passengers")
```

```python
pv
```

```python

import pandas as pd
pd.melt(
    pv,                # DataFrame to reshape
    id_vars="year",         # Columns to keep fixed (not melted)
    value_vars=[1949,1950,1951,1952,1953,1954,1955,1956,1957,1958,1959,1960 ],   # Columns to unpivot (default: all except id_vars)
    var_name="year",        # Name of the new ‘variable’ column
    value_name='passengrs',   # Name of the new ‘value’ column
    
)
```

```python
import pandas as pd

# Convert column names to strings if they are not already
pv.columns = pv.columns.map(str)

# Melt DataFrame
df_melted = pd.melt(
    pv,
    id_vars="month",                      # Column to keep fixed
    value_vars=["1949","1950","1951","1952","1953","1954",
                "1955","1956","1957","1958","1959","1960"], # Must be strings
    var_name="year_column",              # Avoid conflict with 'year' column
    value_name='passengers'              # Fix typo here
)

print(df_melted.head())
```

```python
import matplotlib.pyplot as plt
flights = sns.load_dataset('flights')
pivot_table = flights.pivot(index="month",columns="year", values="passengers")

sns.heatmap(pivot_table, annot=True, fmt="d", cmap="YlGnBu")
plt.title('Heatmap of Passengers')
plt.show()
```
