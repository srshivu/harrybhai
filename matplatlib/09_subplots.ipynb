```python
import matplotlib.pyplot as plt
import numpy as np
x=[1,2,3,4,5]
y1=[3,4,5,6,6]
y2=np.random.randint(5,50,size=5)
color=["red" if i>=2 else "black" for i in range(len(x))]
plt.subplot=(1,2,1)
plt.figure(figsize=(2,3))  # width=8 inches, height=6 inches

plt.scatter(x,y1,color=color,label="red")
plt.xlabel("no")
plt.ylabel("no")
plt.tight_layout()
plt.show()
plt.subplot=(1,2,2)
plt.figure(figsize=(2,3))  # width=8 inches, height=6 inches

plt.scatter(x,y2,color=color,label="red")
plt.xlabel("no")
plt.ylabel("no")
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

exam_score = [45, 67, 89, 30, 75]
students = [1, 2, 3, 4, 5]
colors = ["red" if s >= 60 else "blue" for s in exam_score]

plt.figure(figsize=(10,5))

# 1st subplot (scatter)
plt.subplot(1, 2, 1)  # 1 row, 2 columns, 1st plot
plt.scatter(students, exam_score, c=colors, s=100)
plt.axhline(60, color='gray', linestyle='--')
plt.title('Scatter Plot')
plt.xlabel('Student')
plt.ylabel('Exam Score')

# 2nd subplot (bar)
plt.subplot(1, 2, 2)  # 1 row, 2 columns, 2nd plot
plt.bar(students, exam_score, color=colors)
plt.axhline(60, color='gray', linestyle='--')
plt.title('Bar Chart')
plt.xlabel('Student')
plt.ylabel('Exam Score')

plt.tight_layout()
plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10,5))

# Scatter plot
ax1.scatter(students, exam_score, c=colors, s=100)
ax1.axhline(60, color='gray', linestyle='--')
ax1.set_title('Scatter Plot')
ax1.set_xlabel('Student')
ax1.set_ylabel('Exam Score')

# Bar chart
ax2.bar(students, exam_score, color=colors)
ax2.axhline(60, color='gray', linestyle='--')
ax2.set_title('Bar Chart')
ax2.set_xlabel('Student')
ax2.set_ylabel('Exam Score')

plt.tight_layout()
plt.show()
```
