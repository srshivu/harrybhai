```python
import pandas as pd

# Define a function to apply
def my_function(x):
    return x * 2

# Example DataFrame
df = pd.DataFrame({'A': [1, 2, 3], 'B': [10, 20, 30]})

# Apply the function to each column (default axis=0)
result_col = df.apply(my_function)

# Apply the function to each row (axis=1)
result_row = df.apply(my_function, axis=0)

print("Apply function to each column:")
print(result_col)
print("\nApply function to each row:")
print(result_row)
```
