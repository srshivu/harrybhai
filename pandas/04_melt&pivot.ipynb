```python
import pandas as pd

data = {
    "Student_ID": ["S001", "S002", "S003", "S004", "S005"],
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva"],
    "Math": [85, 67, 90, 72, 95],
    "Science": [78, 88, 82, 69, 91],
    "English": [92, 76, 85, 80, 89],
    "History": [74, 81, 89, 77, 94]
}

df = pd.DataFrame(data)
print("Original DataFrame:\n", df)
```

```python
pd.melt(df,id_vars="Student_ID",value_vars=["Math","Science","English","History"],value_name="score",var_name="sub")
```

```python
df2=pd.melt(df,id_vars=["Student_ID","Name"],value_vars=["Math","Science","English","History"],var_name="sub",value_name="score")
```

```python
df2.sort_values("Student_ID",inplace=True)
```

```python

df2.reset_index()
df2.columns=["name","Student_ID","sub","score"]
```
