```python
import pandas as pd

# DataFrame 1
df1 = pd.DataFrame({
    "EmployeeID": [1, 2, 3],
    "Name": ["Alice", "Bob", "Charlie"]
})

# DataFrame 2
df2 = pd.DataFrame({
    "EmployeeID": [1, 2, 4],
    "Department": ["HR", "Finance", "IT"]
})

# Merge on EmployeeID
merged = pd.merge(df1, df2, on="EmployeeID", how="inner")
print(merged)
```

```python
pd.merge(df1,df2,on="EmployeeID",how="outer")
```

```python
merged=pd.merge(df1,df2,on="EmployeeID",how="left")#it will add all coloms to the left side and match all side of the coloms
print(merged)
```
