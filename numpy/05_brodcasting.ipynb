he ability of NumPy to perform arithmetic operations on arrays of different shapes by automatically stretching (replicating) the smaller array so that both arrays match in shape, without actually copying data.

```python
import numpy as np
arr=np.array([1,2,3,4,5])
add=arr+56
print(add)
```

```python
import numpy as np

arr = np.array([1,2,3,4,5])
king = arr > 3
result = arr[king]+5   # sum of elements > 3
print(result)
```
